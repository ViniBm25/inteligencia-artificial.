```python
#21. Busca Binária | Encontre um elemento em uma lista ordenada.
#-----------------------------------------------------------------------------------------

# Código:

def busca_binaria(lista, valor):
    inicio = 0
    fim = len(lista) - 1

    # Dividi a área de busca pela metade.
    while inicio <= fim:
        meio = (inicio + fim) // 2

        if lista[meio] == valor:
            return meio
        elif lista[meio] < valor:
            inicio = meio + 1
        else:
            fim = meio - 1

    return -1


numeros = [2, 4, 6, 8, 10, 12, 14]
valor = int(input("Digite o número que deseja procurar: "))

posicao = busca_binaria(numeros, valor)

if posicao != -1:
    print("Número encontrado na posição:", posicao)
else:
    print("Número não encontrado.")

#----------------------------------------------------------------------------------------

# Explicação:

# Dividi a lista pela metade em cada repetição até encontrar o valor ou terminar a busca.
```

```python
#22. Criptografia de Cifra de César | Crie uma função para criptografar uma mensagem usando a cifra de César.
#--------------------------------------------------------------------------------------------------

# Código:

# Criei uma função que recebe a mensagem e a quantidade de deslocamento.
def criptografar(mensagem, deslocamento):
    resultado = ""

    # Percorri cada caractere da mensagem.
    for caractere in mensagem:
        if caractere.isalpha():

            # Verifiquei se a letra era maiúscula ou minúscula.
            inicio = ord("A") if caractere.isupper() else ord("a")

            # Calculei a nova posição da letra.
            posicao = (ord(caractere) - inicio + deslocamento) % 26
            resultado += chr(inicio + posicao)
        else:

            # Mantive espaços, números e sinais sem alteração.
            resultado += caractere

    return resultado


mensagem = input("Digite uma mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

print("Mensagem criptografada:", criptografar(mensagem, deslocamento))

#--------------------------------------------------------------------------------------------------

# Explicação:

# Percorri cada letra da mensagem e alterei sua posição de acordo com o deslocamento informado.
# Mantive os espaços, números e sinais e, no final, mostrei a mensagem criptografada.
```

```python
#23. Descriptografar Cifra de César | Recupere uma mensagem cifrada.
#--------------------------------------------------------------------------------------------------

# Código:

def descriptografar(mensagem, deslocamento):
    resultado = ""

    # Retornei cada letra para sua posição original.
    for caractere in mensagem:
        if caractere.isalpha():
            inicio = ord("A") if caractere.isupper() else ord("a")
            posicao = (ord(caractere) - inicio - deslocamento) % 26
            resultado += chr(inicio + posicao)
        else:
            resultado += caractere

    return resultado


mensagem = input("Digite a mensagem criptografada: ")
deslocamento = int(input("Digite o deslocamento utilizado: "))

print("Mensagem original:", descriptografar(mensagem, deslocamento))

#-------------------------------------------------------------------------------------------------

# Explicação:

# Subtraí o deslocamento de cada letra para recuperar a mensagem original.
```

```python
#24. Validação de CPF | Implemente uma função para validar um CPF brasileiro.
#--------------------------------------------------------------------------------------------------

# Código:

def validar_cpf(cpf):
    # Mantive somente os números do CPF.
    cpf = "".join(numero for numero in cpf if numero.isdigit())

    if len(cpf) != 11 or cpf == cpf[0] * 11:
        return False

    # Calculei os dois dígitos verificadores.
    for tamanho in (9, 10):
        soma = 0

        for posicao in range(tamanho):
            soma += int(cpf[posicao]) * (tamanho + 1 - posicao)

        digito = (soma * 10) % 11

        if digito == 10:
            digito = 0

        if digito != int(cpf[tamanho]):
            return False

    return True


cpf = input("Digite um CPF: ")

if validar_cpf(cpf):
    print("CPF válido.")
else:
    print("CPF inválido.")

#--------------------------------------------------------------------------------------------------

# Explicação:

# Retirei a pontuação, verifiquei o tamanho e calculei os dois dígitos verificadores do CPF.
```

```python
#30. Jogo da Forca | Desenvolva um jogo simples para adivinhar uma palavra.
#----------------------------------------------------------------------------------------------

# Código:

import random

palavras = ["python", "computador", "programacao", "faculdade"]

palavra_secreta = random.choice(palavras)
letras_certas = set()
letras_erradas = set()
tentativas = 6

while tentativas > 0:
    palavra_exibida = ""

    for letra in palavra_secreta:
        if letra in letras_certas:
            palavra_exibida += letra + " "
        else:
            palavra_exibida += "_ "

    print("\nPalavra:", palavra_exibida)
    print("Letras erradas:", letras_erradas)
    print("Tentativas restantes:", tentativas)

    if "_" not in palavra_exibida:
        print("Você descobriu a palavra!")
        break

    letra = input("Digite uma letra: ").lower()

    if len(letra) != 1 or not letra.isalpha():
        print("Digite somente uma letra.")
    elif letra in letras_certas or letra in letras_erradas:
        print("Essa letra já foi utilizada.")
    elif letra in palavra_secreta:
        letras_certas.add(letra)
    else:
        letras_erradas.add(letra)
        tentativas -= 1

if tentativas == 0:
    print("Você perdeu. A palavra era:", palavra_secreta)

#---------------------------------------------------------------------------------------------_

# Explicação:

# Sorteei uma palavra e registrei as letras certas e erradas até o acerto ou o fim das tentativas.
```

```python
#31. Ordenação por Bolha | Implemente o algoritmo Bubble Sort.
#----------------------------------------------------------------------------------------
# Código:

def bubble_sort(lista):
    lista = lista.copy()

    # Comparei os números vizinhos.
    for fim in range(len(lista) - 1, 0, -1):
        for posicao in range(fim):
            if lista[posicao] > lista[posicao + 1]:
                lista[posicao], lista[posicao + 1] = (
                    lista[posicao + 1],
                    lista[posicao]
                )

    return lista


numeros = [7, 3, 9, 1, 5]

print("Lista original:", numeros)
print("Lista ordenada:", bubble_sort(numeros))

#---------------------------------------------------------------------------------------

# Explicação:

# Comparei valores vizinhos e troquei suas posições até a lista ficar ordenada.
```

```python
#32. Jogo da Velha | Crie um jogo da velha para dois jogadores.
#----------------------------------------------------------------------------------------------

# Código:

tabuleiro = [" " for posicao in range(9)]


# Criei uma função para mostrar o tabuleiro.
def mostrar_tabuleiro():
    print(f" {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]} ")
    print("---+---+---")
    print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]} ")
    print("---+---+---")
    print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]} ")


# Criei uma função para verificar as combinações vencedoras.
def verificar_vitoria(jogador):
    combinacoes = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    for combinacao in combinacoes:
        if all(tabuleiro[posicao] == jogador for posicao in combinacao):
            return True

    return False


jogador = "X"

for rodada in range(9):
    mostrar_tabuleiro()

    while True:
        try:
            posicao = int(
                input(f"Jogador {jogador}, escolha uma posição de 1 a 9: ")
            ) - 1

            if 0 <= posicao <= 8 and tabuleiro[posicao] == " ":
                break

            print("Posição inválida ou ocupada.")
        except ValueError:
            print("Digite um número de 1 a 9.")

    tabuleiro[posicao] = jogador

    if verificar_vitoria(jogador):
        mostrar_tabuleiro()
        print(f"O jogador {jogador} venceu!")
        break

    jogador = "O" if jogador == "X" else "X"
else:
    mostrar_tabuleiro()
    print("O jogo terminou empatado.")

#----------------------------------------------------------------------------------------------

# Explicação:

# Criei o tabuleiro, alternei os jogadores e verifiquei as combinações de vitória após cada jogada.
```

```python

#35. Números Fibonacci Até N | Gere a sequência até o limite informado.
#--------------------------------------------------------------------------------------------

# Código:

limite = int(input("Gerar Fibonacci até qual número? "))

primeiro = 0
segundo = 1
sequencia = []

while primeiro <= limite:
    sequencia.append(primeiro)
    primeiro, segundo = segundo, primeiro + segundo

print("Sequência de Fibonacci:", sequencia)

#---------------------------------------------------------------------------------------------

# Explicação:

# Somei os dois termos anteriores e continuei até alcançar o limite informado.
```

```python
#37. Funções Recursivas | Calcule o fatorial usando uma função recursiva.
#------------------------------------------------------------------------------------------------

# Código:

def calcular_fatorial(numero):
    # Encerrei as chamadas quando o valor chegou a 0 ou 1.
    if numero == 0 or numero == 1:
        return 1

    return numero * calcular_fatorial(numero - 1)


numero = int(input("Digite um número inteiro não negativo: "))

if numero < 0:
    print("Não existe fatorial de número negativo.")
else:
    print("Resultado do fatorial:", calcular_fatorial(numero))

#--------------------------------------------------------------------------------------------------

# Explicação:

# Fiz a função chamar a si mesma com um número menor até chegar ao caso de encerramento.
```

```python
#40. Lista de Compras | Permita adicionar, remover e listar itens.
#------------------------------------------------------------------------------------------------
# Código:

lista_compras = []

while True:
    print("\n1 - Adicionar item")
    print("2 - Remover item")
    print("3 - Mostrar lista")
    print("4 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        item = input("Digite o item que deseja adicionar: ")
        lista_compras.append(item)
        print("Item adicionado.")

    elif opcao == "2":
        item = input("Digite o item que deseja remover: ")

        if item in lista_compras:
            lista_compras.remove(item)
            print("Item removido.")
        else:
            print("Esse item não está na lista.")

    elif opcao == "3":
        if len(lista_compras) == 0:
            print("A lista está vazia.")
        else:
            print("\nLista de compras:")

            for numero, item in enumerate(lista_compras, start=1):
                print(f"{numero} - {item}")

    elif opcao == "4":
        print("Programa encerrado.")
        break

    else:
        print("Opção inválida.")

#-----------------------------------------------------------------------------------------------

# Explicação:

# Criei um menu que continua funcionando até a opção de sair e trata cada ação escolhida.
```

```python
#42. Números Primos até N | Gere todos os números primos até um limite.
#------------------------------------------------------------------------------------------------

# Código:

def verificar_primo(numero):
    if numero < 2:
        return False

    for divisor in range(2, int(numero ** 0.5) + 1):
        if numero % divisor == 0:
            return False

    return True


limite = int(input("Mostrar números primos até: "))
numeros_primos = []

for numero in range(2, limite + 1):
    if verificar_primo(numero):
        numeros_primos.append(numero)

print("Números primos:", numeros_primos)

#------------------------------------------------------------------------------------------------

# Explicação:

# Verifiquei cada número até o limite e adicionei à lista somente aqueles que eram primos.
```

```python
#46. Calcular Média e Desvio Padrão | Calcule as duas medidas de uma lista.
#-------------------------------------------------------------------------------------------------

# Código:

import math

numeros = [10, 12, 15, 18, 20]

# Calculei a média.
media = sum(numeros) / len(numeros)

soma_diferencas = 0

for numero in numeros:
    soma_diferencas += (numero - media) ** 2

# Calculei a variância e o desvio padrão.
variancia = soma_diferencas / len(numeros)
desvio_padrao = math.sqrt(variancia)

print(f"Média: {media:.2f}")
print(f"Desvio padrão: {desvio_padrao:.2f}")

#----------------------------------------------------------------------------------------------

# Explicação:

# Calculei a média, a variância e a raiz quadrada da variância para obter o desvio padrão.
```

```python
#47. Contar Números em Lista | Conte quantas vezes cada número aparece.
#-----------------------------------------------------------------------------------------------

# Código:

numeros = [1, 2, 2, 3, 1, 4, 2, 3]
contagem = {}

# Guardei cada número e sua quantidade em um dicionário.
for numero in numeros:
    if numero in contagem:
        contagem[numero] += 1
    else:
        contagem[numero] = 1

for numero, quantidade in contagem.items():
    print(f"O número {numero} aparece {quantidade} vez(es).")

#-----------------------------------------------------------------------------------------------

# Explicação:

# Usei um dicionário para relacionar cada número à quantidade de vezes em que apareceu.
```

```python
#48. Ordenação por Inserção | Implemente o algoritmo Insertion Sort.
#-----------------------------------------------------------------------------------------------

# Código:

def insertion_sort(lista):
    lista = lista.copy()

    for posicao_atual in range(1, len(lista)):
        numero_atual = lista[posicao_atual]
        posicao_anterior = posicao_atual - 1

        # Desloquei os valores maiores para abrir espaço.
        while (
            posicao_anterior >= 0
            and lista[posicao_anterior] > numero_atual
        ):
            lista[posicao_anterior + 1] = lista[posicao_anterior]
            posicao_anterior -= 1

        lista[posicao_anterior + 1] = numero_atual

    return lista


numeros = [8, 4, 2, 9, 5, 1]

print("Lista original:", numeros)
print("Lista ordenada:", insertion_sort(numeros))

#--------------------------------------------------------------------------------------------------

# Explicação:

# Analisei cada elemento e o inseri na posição correta entre os valores anteriores.
```

```python
#50. Cálculo de Frequência de Letras | Calcule a frequência de cada letra.
#--------------------------------------------------------------------------------------------------

# Código:

texto = input("Digite um texto: ").lower()
frequencia = {}

# Contei somente os caracteres que eram letras.
for letra in texto:
    if letra.isalpha():
        if letra in frequencia:
            frequencia[letra] += 1
        else:
            frequencia[letra] = 1

for letra in sorted(frequencia):
    print(f"A letra '{letra}' aparece {frequencia[letra]} vez(es).")

#--------------------------------------------------------------------------------------------------

# Explicação:

# Percorri o texto e usei um dicionário para guardar a frequência de cada letra.
```
