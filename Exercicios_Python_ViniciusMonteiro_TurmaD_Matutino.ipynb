```python
# Exercício 21
# Divido a área de busca da lista pela metade até encontrar o número.
# Escolhi a busca binária porque ela evita verificar todos os elementos,
# mas só funciona corretamente quando a lista está ordenada.

def busca_binaria(lista, valor):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2

        if lista[meio] == valor:
            return meio
        elif lista[meio] < valor:
            inicio = meio + 1
        else:
            fim = meio - 1

    return -1


numeros = [2, 4, 6, 8, 10, 12, 14]
valor = int(input("Digite o número que deseja procurar: "))

posicao = busca_binaria(numeros, valor)

if posicao != -1:
    print("Número encontrado na posição:", posicao)
else:
    print("Número não encontrado.")
```

```python
# Exercício 22
# Desloco cada letra da mensagem pela quantidade informada para criptografá-la.
# Uso uma função para organizar o processo e permitir que outras mensagens
# sejam criptografadas com diferentes deslocamentos.

def criptografar(mensagem, deslocamento):
    resultado = ""

    for caractere in mensagem:
        if caractere.isalpha():
            inicio = ord("A") if caractere.isupper() else ord("a")
            posicao = (ord(caractere) - inicio + deslocamento) % 26
            resultado += chr(inicio + posicao)
        else:
            resultado += caractere

    return resultado


mensagem = input("Digite uma mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

print("Mensagem criptografada:", criptografar(mensagem, deslocamento))
```

```python
# Exercício 23
# Desloco as letras no sentido contrário para recuperar a mensagem original.
# Uso uma função parecida com a da criptografia, mas subtraio o deslocamento
# em vez de somar.

def descriptografar(mensagem, deslocamento):
    resultado = ""

    for caractere in mensagem:
        if caractere.isalpha():
            inicio = ord("A") if caractere.isupper() else ord("a")
            posicao = (ord(caractere) - inicio - deslocamento) % 26
            resultado += chr(inicio + posicao)
        else:
            resultado += caractere

    return resultado


mensagem = input("Digite a mensagem criptografada: ")
deslocamento = int(input("Digite o deslocamento utilizado: "))

print("Mensagem original:", descriptografar(mensagem, deslocamento))
```

```python
# Exercício 24
# Retiro os pontos e o traço do CPF e calculo os dois dígitos verificadores.
# Uso uma função para reunir todas as etapas da validação e retornar
# se o CPF informado é válido ou inválido.

def validar_cpf(cpf):
    cpf = "".join(numero for numero in cpf if numero.isdigit())

    if len(cpf) != 11 or cpf == cpf[0] * 11:
        return False

    for tamanho in (9, 10):
        soma = 0

        for posicao in range(tamanho):
            soma += int(cpf[posicao]) * (tamanho + 1 - posicao)

        digito = (soma * 10) % 11

        if digito == 10:
            digito = 0

        if digito != int(cpf[tamanho]):
            return False

    return True


cpf = input("Digite um CPF: ")

if validar_cpf(cpf):
    print("CPF válido.")
else:
    print("CPF inválido.")
```

```python
import random

# Exercício 30
# Sorteio uma palavra e guardo separadamente as letras certas e erradas.
# Uso um laço while para manter o jogo funcionando até o jogador completar
# a palavra ou atingir o limite de erros.

palavras = ["python", "computador", "programacao", "faculdade"]

palavra_secreta = random.choice(palavras)
letras_certas = set()
letras_erradas = set()
tentativas = 6

while tentativas > 0:
    palavra_exibida = ""

    for letra in palavra_secreta:
        if letra in letras_certas:
            palavra_exibida += letra + " "
        else:
            palavra_exibida += "_ "

    print("\nPalavra:", palavra_exibida)
    print("Letras erradas:", letras_erradas)
    print("Tentativas restantes:", tentativas)

    if "_" not in palavra_exibida:
        print("Você descobriu a palavra!")
        break

    letra = input("Digite uma letra: ").lower()

    if len(letra) != 1 or not letra.isalpha():
        print("Digite somente uma letra.")
    elif letra in letras_certas or letra in letras_erradas:
        print("Essa letra já foi utilizada.")
    elif letra in palavra_secreta:
        letras_certas.add(letra)
    else:
        letras_erradas.add(letra)
        tentativas -= 1

if tentativas == 0:
    print("Você perdeu. A palavra era:", palavra_secreta)
```

```python
# Exercício 31
# Comparo os números vizinhos e troco suas posições quando estão fora de ordem.
# Repito esse processo porque o exercício pede o Bubble Sort, que leva
# os maiores valores para o final da lista aos poucos.

def bubble_sort(lista):
    lista = lista.copy()

    for fim in range(len(lista) - 1, 0, -1):
        for posicao in range(fim):
            if lista[posicao] > lista[posicao + 1]:
                lista[posicao], lista[posicao + 1] = (
                    lista[posicao + 1],
                    lista[posicao]
                )

    return lista


numeros = [7, 3, 9, 1, 5]

print("Lista original:", numeros)
print("Lista ordenada:", bubble_sort(numeros))
```

```python
# Exercício 32
# Crio um tabuleiro para dois jogadores escolherem posições alternadamente.
# Uso funções para organizar a exibição e a verificação da vitória,
# deixando cada parte do jogo separada e mais fácil de entender.

tabuleiro = [" " for posicao in range(9)]


def mostrar_tabuleiro():
    print(f" {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]} ")
    print("---+---+---")
    print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]} ")
    print("---+---+---")
    print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]} ")


def verificar_vitoria(jogador):
    combinacoes = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    for combinacao in combinacoes:
        if all(tabuleiro[posicao] == jogador for posicao in combinacao):
            return True

    return False


jogador = "X"

for rodada in range(9):
    mostrar_tabuleiro()

    while True:
        try:
            posicao = int(
                input(f"Jogador {jogador}, escolha uma posição de 1 a 9: ")
            ) - 1

            if 0 <= posicao <= 8 and tabuleiro[posicao] == " ":
                break

            print("Posição inválida ou ocupada.")
        except ValueError:
            print("Digite um número de 1 a 9.")

    tabuleiro[posicao] = jogador

    if verificar_vitoria(jogador):
        mostrar_tabuleiro()
        print(f"O jogador {jogador} venceu!")
        break

    jogador = "O" if jogador == "X" else "X"
else:
    mostrar_tabuleiro()
    print("O jogo terminou empatado.")
```

```python
# Exercício 40
# Crio um menu para adicionar, remover e mostrar os itens da lista de compras.
# Uso o while porque o menu deve continuar aparecendo até o usuário
# escolher a opção de sair.

lista_compras = []

while True:
    print("\n1 - Adicionar item")
    print("2 - Remover item")
    print("3 - Mostrar lista")
    print("4 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        item = input("Digite o item que deseja adicionar: ")
        lista_compras.append(item)
        print("Item adicionado.")

    elif opcao == "2":
        item = input("Digite o item que deseja remover: ")

        if item in lista_compras:
            lista_compras.remove(item)
            print("Item removido.")
        else:
            print("Esse item não está na lista.")

    elif opcao == "3":
        if len(lista_compras) == 0:
            print("A lista está vazia.")
        else:
            print("\nLista de compras:")

            for numero, item in enumerate(lista_compras, start=1):
                print(f"{numero} - {item}")

    elif opcao == "4":
        print("Programa encerrado.")
        break

    else:
        print("Opção inválida.")
```

```python
# Exercício 45
# Verifico o resto da divisão de cada número por 2
# e separo os valores em listas de pares e ímpares.

numeros = [10, 7, 3, 8, 12, 15]

pares = []
impares = []

for numero in numeros:
    if numero % 2 == 0:
        pares.append(numero)
    else:
        impares.append(numero)

print("Números pares:", pares)
print("Números ímpares:", impares)
```

```python
import math


# Exercício 46
# Primeiro calculo a média da lista e depois uso a diferença de cada valor
# para chegar ao resultado do desvio padrão.

numeros = [10, 12, 15, 18, 20]

media = sum(numeros) / len(numeros)

soma_diferencas = 0

for numero in numeros:
    soma_diferencas += (numero - media) ** 2

variancia = soma_diferencas / len(numeros)
desvio_padrao = math.sqrt(variancia)

print(f"Média: {media:.2f}")
print(f"Desvio padrão: {desvio_padrao:.2f}")
```

```python
# Exercício 47
# Uso um dicionário para guardar cada número da lista
# junto da quantidade de vezes que ele apareceu.
numeros = [1, 2, 2, 3, 1, 4, 2, 3]
contagem = {}

for numero in numeros:
    if numero in contagem:
        contagem[numero] += 1
    else:
        contagem[numero] = 1

for numero, quantidade in contagem.items():
    print(f"O número {numero} aparece {quantidade} vez(es).")
```

```python
# Exercício 48
# Comparo cada número com os valores anteriores e o coloco na posição correta.
# Uso esse processo porque o exercício pede o Insertion Sort, que organiza
# a lista inserindo cada elemento no local adequado.

def insertion_sort(lista):
    lista = lista.copy()

    for posicao_atual in range(1, len(lista)):
        numero_atual = lista[posicao_atual]
        posicao_anterior = posicao_atual - 1

        while (
            posicao_anterior >= 0
            and lista[posicao_anterior] > numero_atual
        ):
            lista[posicao_anterior + 1] = lista[posicao_anterior]
            posicao_anterior -= 1

        lista[posicao_anterior + 1] = numero_atual

    return lista


numeros = [8, 4, 2, 9, 5, 1]

print("Lista original:", numeros)
print("Lista ordenada:", insertion_sort(numeros))
```

```python
# Exercício 49
# Transformo o número em texto para passar por cada dígito
# e somar todos eles no final.

numero = int(input("Digite um número inteiro: "))
numero_positivo = abs(numero)
soma = 0

for digito in str(numero_positivo):
    soma += int(digito)

print("A soma dos dígitos é:", soma)
```

```python
# Exercício 50
# Analiso somente as letras do texto e guardo em um dicionário
# a quantidade de vezes que cada uma apareceu.

texto = input("Digite um texto: ").lower()
frequencia = {}

for letra in texto:
    if letra.isalpha():
        if letra in frequencia:
            frequencia[letra] += 1
        else:
            frequencia[letra] = 1

for letra in sorted(frequencia):
    print(f"A letra '{letra}' aparece {frequencia[letra]} vez(es).")
```
